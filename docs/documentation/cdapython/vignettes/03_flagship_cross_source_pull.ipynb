# Trusting a cross-source cohort: checking data freshness and saving results

*Part 3 of 3. Picks up from [Part 2](#), where I built a cohort of female subjects with any diagnosis touching the circulatory system, harmonized across all six CDA sources. This part stands on its own if you're starting here -- the only assumption is a `get_subject_data()` result in hand.*

I have my cohort. Before I trust it for anything, I want to see how much each source actually contributed, and whether that breakdown reflects real differences in the data -- or just differences in how recently each source was refreshed.

```python
from cdapython import get_subject_data

cohort = get_subject_data(
    match_all=['sex = F*', 'anatomic_site = circulatory system'],
    data_source=['GDC', 'PDC', 'IDC', 'GC', 'ICDC', 'CTDC'],
    add_extras=['all']
)
cohort['data_source'].value_counts()
```

Before I read anything into these numbers -- a source with fewer matching rows isn't necessarily under-represented in reality. It might just mean that source's data was extracted less recently than the others, so it simply hasn't been refreshed with as much matching data yet. Before drawing any conclusion from this breakdown, I check the [data currency page](#) for each source's last extraction date.

Satisfied with that, I save the cohort out rather than keep working with it only in-session:

```python
get_subject_data(
    match_all=['sex = F*', 'anatomic_site = circulatory system'],
    data_source=['GDC', 'PDC', 'IDC', 'GC', 'ICDC', 'CTDC'],
    add_extras=['all'],
    return_data_as='tsv',
    output_file='circulatory_cohort.tsv'
)
```

A matching `get_file_data()` call takes the same arguments, if I need file-level results instead of subject-level for this cohort.
